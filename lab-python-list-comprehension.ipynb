````python
# 1. Define list of products
products = ["t-shirt", "mug", "hat", "book", "keychain"]


# 2 & 3. MODIFIED: Initialize inventory using Dictionary Comprehension
def initialize_inventory(products):
    inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
    return inventory


# 4 & 5. MODIFIED: Get customer orders using List Comprehension based on a fixed count
def get_customer_orders():
    # Replaces the old dynamic 'while' loop with an explicit count-based entry
    num_orders = int(input("Enter the number of customer orders: "))
    customer_orders = [input("Enter the name of a product that a customer wants to order: ").strip().lower() for _ in range(num_orders)]
    return customer_orders


# NEW FUNCTION: Prompts for individual prices and computes the cumulative sum
def calculate_total_price(customer_orders):
    # Uses a list comprehension to sequentially prompt for pricing metrics
    prices = [float(input(f"Enter the price of {product}: ")) for product in customer_orders]
    return sum(prices)


# 9. MODIFIED: Decrements stock levels and filters out items that reach 0 quantity
def update_inventory(inventory, customer_orders):
    # First, subtract 1 for each ordered item (Maintains original loop logic)
    for product in customer_orders:
        if product in inventory:
            inventory[product] -= 1
            
    # DICTIONARY COMPREHENSION: Filters out and drops any products with a quantity of zero
    updated_inventory = {product: quantity for product, quantity in inventory.items() if quantity > 0}
    return updated_inventory


# --- EXECUTION FLOW ---

# Run inventory initialization setup
inventory = initialize_inventory(products)

# Capture dynamic incoming customer purchases
customer_orders = get_customer_orders()

# 7 & 8. Calculate statistics (Updated to compute absolute unique percentages)
total_products_ordered = len(customer_orders)
percentage_unique = (len(set(customer_orders)) / len(products)) * 100

print("\nOrder Statistics:")
print(f"Total Products Ordered: {total_products_ordered}")
print(f"Percentage of Unique Products Ordered: {percentage_unique:.1f}")

# 9 & 10. Update and display the remaining stock balances
inventory = update_inventory(inventory, customer_orders)

print("\nUpdated Inventory:")
for product, quantity in inventory.items():
    print(f"{product}: {quantity}")

# NEW STEP: Process price calculation prompts and output the final statement total
total_price = calculate_total_price(customer_orders)
print(f"Total Price: {total_price:.1f}")

````
